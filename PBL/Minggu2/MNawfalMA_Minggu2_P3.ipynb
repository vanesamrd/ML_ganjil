# Minggu 2: Praktikum 3 - Scaling & Encoding


```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

df = pd.read_csv('../datasetPBL/Transactions.csv')

# 1. Scaling NetSales
scaler = StandardScaler()
df['NetSales_Scaled'] = scaler.fit_transform(df[['NetSales']])

# 2. Encoding PaymentMethod
le = LabelEncoder()
df['PaymentMethod_Encoded'] = le.fit_transform(df['PaymentMethod'])

print("Hasil Scaling & Encoding (5 data teratas):")
display(df[['NetSales', 'NetSales_Scaled', 'PaymentMethod', 'PaymentMethod_Encoded']].head())
```
