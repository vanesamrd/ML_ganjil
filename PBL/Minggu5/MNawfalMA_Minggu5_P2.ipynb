# Minggu 5: Praktikum 2 - Evaluasi Klaster (Silhouette Score)


```python
import pandas as pd
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('../datasetPBL/Transactions.csv').sample(1000, random_state=42)
X = df[['NetSales', 'TotalAmount']]
X_scaled = StandardScaler().fit_transform(X)

# K-Means
km = KMeans(n_clusters=3).fit(X_scaled)
score_km = silhouette_score(X_scaled, km.labels_)

# Agglomerative
hc = AgglomerativeClustering(n_clusters=3).fit(X_scaled)
score_hc = silhouette_score(X_scaled, hc.labels_)

print(f"Silhouette K-Means: {score_km:.3f}")
print(f"Silhouette Agglomerative: {score_hc:.3f}")
```
