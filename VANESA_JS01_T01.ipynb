2. Dampak Energi dan Lingkungan dari Pemanfaatan Kecerdasan Buatan serta Cara Mengatasinya
Dampak terhadap energi dan lingkungan:
Pelatihan dan operasional model AI berskala besar (terutama large language models) membutuhkan daya komputasi sangat besar, sehingga meningkatkan konsumsi listrik data center secara signifikan. Riset memperkirakan beban kerja AI akan menyumbang lebih dari 50% konsumsi daya data center pada tahun 2028.
Saat ini data center diperkirakan menggunakan sekitar 1,5% listrik global dan berkontribusi sekitar 1% emisi gas rumah kaca terkait energi, dengan tren yang terus meningkat seiring adopsi AI.
Sistem pendingin server (cooling system) pada data center membutuhkan konsumsi air dalam jumlah besar.
Produksi perangkat keras (GPU/chip AI) melibatkan penambangan mineral langka dan menghasilkan limbah elektronik (e-waste) di akhir siklus hidup perangkat.
Jika sumber energi yang digunakan data center masih didominasi bahan bakar fosil, jejak karbon dari operasional AI menjadi semakin besar.
Cara mengatasinya:
Menerapkan green data center, yaitu memanfaatkan energi terbarukan (surya, angin, air) untuk menyuplai kebutuhan listrik data center, serta teknologi pendingin hemat energi seperti immersion cooling.
Menerapkan praktik green software engineering, misalnya mengoptimalkan algoritma dan arsitektur model agar tidak boros komputasi (contoh: model compression, distillation, quantization).
Menjalankan beban kerja komputasi AI secara carbon-aware, yaitu pada lokasi dan waktu tertentu yang bertepatan dengan ketersediaan energi terbarukan.
Menerapkan standar manajemen lingkungan seperti ISO 14001 pada operasional data center, termasuk pemantauan metrik seperti Power Usage Effectiveness (PUE) dan Water Usage Effectiveness (WUE).
Mendaur ulang perangkat keras dan komponen elektronik untuk mengurangi limbah elektronik (e-waste).Pendapat pribadi:
Menurut saya, dilema utama AI dan lingkungan adalah AI di satu sisi menjadi penyebab lonjakan konsumsi energi, tetapi di sisi lain AI juga bisa menjadi bagian dari solusinya — misalnya digunakan untuk mengoptimalkan pendinginan data center secara otomatis (seperti studi kasus data center Facebook di Luleå, Swedia, yang berhasil menekan konsumsi energi secara drastis menggunakan AI). Karena itu, pengembang dan pengguna AI perlu menerapkan prinsip efficiency by design sejak awal — bukan hanya mengejar akurasi atau kecepatan model, tapi juga mempertimbangkan efisiensi energi sebagai bagian dari kriteria keberhasilan sebuah sistem AI.
